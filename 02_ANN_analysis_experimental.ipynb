```python
#importing the stuff that will be needed for the analysis
import pandas as pd
from sklearn.metrics import confusion_matrix
import seaborn as sns 
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
```

```python
#importing pandas to read the csv
predictions = pd.read_csv('predictions_binary.csv')
```

```python
#converting them into values
y_test = predictions["diabetes"].values
y_prob = predictions["diabetes_pred"].values
```

These are the experimental and experiment 4 is the optimal that we chose to use.

Experiment 1:

```python
#setting the threshold for the confusion matrix 
y_pred_binary1 = (y_prob > 0.2).astype(int)

#generate heatmap for the confusion matrix using seaborn

sns.heatmap((confusion_matrix(y_test, y_pred_binary1, labels=[1, 0])).T, cmap="Blues", annot=True, fmt='g', 
    cbar_kws={"orientation":"vertical", "label":"color bar"},xticklabels=["Yes Diabetes", "No Diabetes"],
    yticklabels=["Yes Diabetes", "No Diabetes"])
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.show()
```

```python
#classification report giving metrics like sensitivity and specificity
print(classification_report(y_test, y_pred_binary1, target_names=['No Diabetes','Diabetes']))
```

```python
#Metrics summary for the overall analysis put into a frame

tn, fp, fn, tp = confusion_matrix(y_test, y_pred_binary1).ravel()
specificity = tn / (tn + fp)
sensitivity = tp / (tp + fn)

metrics_summary = {
    'Accuracy': accuracy_score(y_test, y_pred_binary1),
    'Precision (No Diabetes)': precision_score(y_test, y_pred_binary1, pos_label=0),
    'F1-Score (No Diabetes)': f1_score(y_test, y_pred_binary1, pos_label=0),
    'Precision (Diabetes)': precision_score(y_test, y_pred_binary1, pos_label=1),
    'F1-Score (Diabetes)': f1_score(y_test, y_pred_binary1, pos_label=1),
    'Sensitivity': sensitivity,
    'Specificity': specificity}

metrics_df = pd.DataFrame([metrics_summary])
print(metrics_df.T)
```

Best metric performance:

```python
#best threshold performance
#better table presented
import numpy as np

report = classification_report(
    y_test, y_pred_binary5,
    target_names=['No Diabetes', 'Yes Diabetes'],
    output_dict=True
)
df = pd.DataFrame(report).T

row_order = ['No Diabetes', 'Yes Diabetes', 'accuracy', 'macro avg', 'weighted avg']
df = df.reindex([r for r in row_order if r in df.index])

num_cols = [c for c in df.columns if c != 'support']
df[num_cols] = df[num_cols].astype(float).round(2)
if 'support' in df.columns:
    df['support'] = df['support'].astype(int)

fig_h = 1.5 + 0.35 * len(df)         
fig_w = 6.8                           
fig, ax = plt.subplots(figsize=(fig_w, fig_h))
ax.axis('off')

table_data = np.column_stack([df.index.to_list(), df.values])
col_labels = ['Class / Avg'] + df.columns.to_list()

tbl = ax.table(cellText=table_data, colLabels=col_labels, loc='center')
tbl.auto_set_font_size(False)
tbl.set_fontsize(9)
tbl.scale(1, 1.2)

plt.tight_layout()
#plt.savefig('02: optimal classification_report 01.png', dpi=200, bbox_inches='tight')
```
